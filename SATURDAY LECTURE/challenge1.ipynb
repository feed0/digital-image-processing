- Draw a polygon around the previous uni logo





## 0 Import

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## Constants

```python
UNI_ENTRANCE_PATH = 'build.jpg'
UNI_ENTRANCE_IMG_BGR = cv2.imread(UNI_ENTRANCE_PATH)
UNI_ENTRANCE_IMG_RGB = cv2.cvtColor(UNI_ENTRANCE_IMG_BGR, cv2.COLOR_BGR2RGB)

RED_COLOR = (255,0,0)
```

## Load images

```python
plt.imshow(UNI_ENTRANCE_IMG_RGB)
```

## Polylines

### vertices

```python
uni_entrance_img_rgb    = UNI_ENTRANCE_IMG_RGB.copy()
vertices                = np.array([[70, 40],
                                    [210, 50],
                                    [209, 190],
                                    [70 ,200]], np.int32)
pts                     = vertices.reshape((-1,1,2))
```

```python
vertices
```

```python
pts
```

### Draw on top

```python
# Draw the polygon on top of the image
cv2.polylines(uni_entrance_img_rgb,
              [pts],
              isClosed = True,
              color = RED_COLOR,
              thickness = 3)

# Plot
plt.imshow(uni_entrance_img_rgb)
```
